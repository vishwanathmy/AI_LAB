```python

import random

board = [' ' for _ in range(9)]



def display_board():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()



def check_winner(player):
    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_positions:
        if board[a] == player and board[b] == player and board[c] == player:
            return True

    return False


def board_full():
    return ' ' not in board



def computer_move():
    empty_positions = []

    for i in range(9):
        if board[i] == ' ':
            empty_positions.append(i)

    position = random.choice(empty_positions)
    board[position] = 'O'

    print("Computer chose position:", position + 1)



print("TIC-TAC-TOE")
print("You = X")
print("Computer = O")

print("\nPosition numbers:")
print("1 | 2 | 3")
print("--+---+--")
print("4 | 5 | 6")
print("--+---+--")
print("7 | 8 | 9")

while True:


    display_board()

    while True:
        try:
            position = int(input("Enter your position (1-9): "))

            if position < 1 or position > 9:
                print("Enter a number between 1 and 9.")
            elif board[position - 1] != ' ':
                print("Position already occupied.")
            else:
                board[position - 1] = 'X'
                break

        except ValueError:
            print("Enter a valid number.")


    if check_winner('X'):
        display_board()
        print("You Win!")
        break

    if board_full():
        display_board()
        print("Game Draw!")
        break


    computer_move()

    if check_winner('O'):
        display_board()
        print("Computer Wins!")
        break


    if board_full():
        display_board()
        print("Game Draw!")
        break



```

```python
room_A = input("Enter status of Room A (Dirty/Clean): ").capitalize()
room_B = input("Enter status of Room B (Dirty/Clean): ").capitalize()

location = input("Enter vacuum starting location (A/B): ").upper()

rooms = {
    'A': room_A,
    'B': room_B
}

print("\nInitial State:")
print(rooms)
print("Vacuum Location:", location)

while True:

    if rooms[location] == "Dirty":

        print("Action: SUCK")
        rooms[location] = "Clean"

    else:

        print("Action: MOVE")

        if location == 'A':
            location = 'B'
        else:
            location = 'A'


    if rooms['A'] == "Clean" and rooms['B'] == "Clean":
        break

print("\nFinal State:")
print(rooms)
print("Vacuum Location:", location)
print("Both rooms are clean.")


```
