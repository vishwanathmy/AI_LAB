```python
initial = (1, 2, 3,
           4, 5, 0,
           7, 8, 6)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def get_neighbors(state):
    neighbors = []

    blank = state.index(0)
    row, col = divmod(blank, 3)

    moves = [(-1, 0), (1, 0),
             (0, -1), (0, 1)]

    for dr, dc in moves:
        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            pos = r * 3 + c
            board = list(state)

            board[blank], board[pos] = (
                board[pos], board[blank]
            )

            neighbors.append(tuple(board))

    return neighbors


def dfs(state, goal, visited, path, limit):
    if state == goal:
        return path

    if limit == 0:
        return None

    for next_state in get_neighbors(state):
        if next_state not in visited:
            visited.add(next_state)

            result = dfs(
                next_state,
                goal,
                visited,
                path + [next_state],
                limit - 1
            )

            if result is not None:
                return result

            visited.remove(next_state)

    return None


def print_solution(solution):
    if solution is None:
        print("No solution found within the depth limit.")
        return

    print("Solution found!")
    print("Moves:", len(solution) - 1)

    for step, state in enumerate(solution):
        print("Step", step)
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()


solution = dfs(initial, goal, {initial}, [initial], 20)
print_solution(solution)


```

```python
initial = (1, 2, 3,
           4, 5, 6,
           7, 0, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def get_neighbors(state):
    neighbors = []

    blank = state.index(0)
    row, col = divmod(blank, 3)

    moves = [(-1, 0), (1, 0),
             (0, -1), (0, 1)]

    for dr, dc in moves:
        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            pos = r * 3 + c
            board = list(state)

            board[blank], board[pos] = (
                board[pos], board[blank]
            )

            neighbors.append(tuple(board))

    return neighbors


def depth_limited_dfs(state, goal, limit, path, visited):
    if state == goal:
        return path

    if limit == 0:
        return None

    for next_state in get_neighbors(state):
        if next_state not in visited:
            visited.add(next_state)

            result = depth_limited_dfs(
                next_state,
                goal,
                limit - 1,
                path + [next_state],
                visited
            )

            if result is not None:
                return result

            visited.remove(next_state)

    return None


def ids(initial, goal, max_depth=30):
    for depth in range(max_depth + 1):
        print("Depth limit:", depth)

        solution = depth_limited_dfs(
            initial, goal, depth, [initial], {initial}
        )

        if solution is not None:
            return solution

    return None


def print_solution(solution):
    if solution is None:
        print("No solution found within the depth limit.")
        return

    print("Solution found!")
    print("Moves:", len(solution) - 1)

    for step, state in enumerate(solution):
        print("Step", step)
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()


solution = ids(initial, goal)
print_solution(solution)
```
