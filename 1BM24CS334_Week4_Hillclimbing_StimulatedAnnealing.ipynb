```python
import random

N = 4

def cost(board):
    """Count pairs of queens attacking each other."""
    conflicts = 0
    for i in range(N):
        for j in range(i + 1, N):
            if (board[i] == board[j] or
                abs(board[i] - board[j]) == abs(i - j)):
                conflicts += 1
    return conflicts


def get_neighbors(board):
    """Move one queen to another row in its column."""
    neighbors = []
    for col in range(N):
        for row in range(1, N + 1):
            if row != board[col]:
                neighbor = board.copy()
                neighbor[col] = row
                neighbors.append((col + 1, row, neighbor))
    return neighbors


def hill_climbing():
    restart = 0

    while True:
        restart += 1
        current = [random.randint(1, N) for _ in range(N)]

        print(f"\nRestart {restart}: Initial state = {current}")
        print("Initial cost =", cost(current))

        while True:
            current_cost = cost(current)

            if current_cost == 0:
                print("\nGOAL FOUND!")
                print("Final board:", current)
                print("Final cost:", current_cost)
                return current

            neighbors = get_neighbors(current)

            print("\nCurrent state:", current)
            print("Current cost:", current_cost)
            print("Neighbor states and costs:")

            for col, row, neighbor in neighbors:
                print(
                    f" Move column {col} to row {row}: "
                    f"{neighbor}, cost = {cost(neighbor)}"
                )

            # Choose a minimum-cost neighbor.
            best_cost = min(cost(nb) for _, _, nb in neighbors)
            best_neighbors = [
                nb for _, _, nb in neighbors
                if cost(nb) == best_cost
            ]
            best = random.choice(best_neighbors)

            # Restart if no improvement is possible.
            if best_cost >= current_cost:
                print("Local minimum reached; restarting...")
                break

            current = best


if __name__ == "__main__":
    hill_climbing()
```

```python

import random
import math

N = 8

def heuristic(board):
    conflicts = 0
    for i in range(N):
        for j in range(i + 1, N):
            if abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1
    return conflicts


def print_board(board):
    for row in range(N):
        print(" ".join(
            "Q" if board[col] == row else "."
            for col in range(N)
        ))


def simulated_annealing():
    # Initial board with a conflict
    board = [0, 4, 7, 5, 2, 6, 1, 1]

    temperature = 10.0
    cooling_rate = 0.95
    step = 0

    print("Initial Board:")
    print(board)
    print_board(board)
    print("Initial Conflicts:", heuristic(board))

    while heuristic(board) != 0 and step < 1000:
        step += 1
        current_cost = heuristic(board)

        col = random.randrange(N)
        old_row = board[col]
        new_row = random.choice(
            [r for r in range(N) if r != old_row]
        )

        board[col] = new_row
        new_cost = heuristic(board)
        delta = new_cost - current_cost

        if delta <= 0:
            accepted = True
        else:
            accepted = random.random() < math.exp(
                -delta / temperature
            )

        if not accepted:
            board[col] = old_row
            new_cost = current_cost

        print(f"\nStep {step}")
        print(f"Move Queen: Column {col + 1}, "
              f"Row {old_row + 1} -> Row "
              f"{board[col] + 1}")
        print(f"Temperature: {temperature:.4f}")
        print(f"Conflicts: {current_cost} -> {new_cost}")
        print("Move:", "Accepted" if accepted else "Rejected")
        print("Current Board:", board)

        temperature *= cooling_rate

    print("\nFinal Result:")
    if heuristic(board) == 0:
        print("Solution Found!")
        print_board(board)
        print("Conflicts:", heuristic(board))
        print("Total Steps:", step)
    else:
        print("Solution not found within 1000 steps.")
        print("Conflicts:", heuristic(board))


simulated_annealing()
```
