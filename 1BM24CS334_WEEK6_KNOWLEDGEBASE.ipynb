```python

def tt_entails(KB, alpha, symbols):
    print("\nTruth Table:")
    print("-" * 45)
    print(" | ".join(symbols) + " | KB | Alpha")
    print("-" * 45)

    result = tt_check_all(KB, alpha, symbols, {})

    print("-" * 45)
    print("KB entails Alpha:", result)


def tt_check_all(KB, alpha, symbols, model):

    if len(symbols) == 0:
        kb_value = KB(model)
        alpha_value = alpha(model)

        values = [str(model[s]) for s in model]
        print(" | ".join(values),
              "|", kb_value, "|", alpha_value)

        if kb_value:
            return alpha_value
        else:
            return True

    P = symbols[0]
    rest = symbols[1:]

    model_true = model.copy()
    model_true[P] = True

    model_false = model.copy()
    model_false[P] = False

    return (tt_check_all(KB, alpha, rest, model_true)
            and
            tt_check_all(KB, alpha, rest, model_false))


# INPUT
KB = lambda m: m['P'] or m['Q']
alpha = lambda m: m['P']
symbols = ['P', 'Q']

# RUN
tt_entails(KB, alpha, symbols)
```
